```python
import pandas as pd
df = pd.read_csv("../exercice-08/data/titanic_propre.csv")
len(df)
```

```python
ports = pd.DataFrame({
    "embark_town": ["Southampton", "Cherbourg", "Queenstown"],
    "pays": ["Angleterre", "France", "Irlande"],
})
ports
```

```python
titanic_pays = df.merge(ports, on="embark_town", how="left")
len(df), len(titanic_pays)
```

```python
effectifs = titanic_pays["pays"].value_counts()
effectifs
```

```python
survie_par_pays = (titanic_pays.groupby("pays")["survived"].mean() * 100).round(1)
survie_par_pays
```

```python
ports_sans_queenstown = ports[ports["embark_town"] != "Queenstown"]

jointure_inner = df.merge(ports_sans_queenstown, on="embark_town", how="inner")
jointure_left = df.merge(ports_sans_queenstown, on="embark_town", how="left")

len(df), len(jointure_inner), len(jointure_left)
```

```python
jointure_left["pays"].isna().sum()
```

### Ce qui se passe quand on retire Queenstown

- Avec `inner`, on passe de 891 à 814 lignes : les 77 passagers de Queenstown disparaissent, sans erreur ni avertissement.
- Avec `left`, on garde 891 lignes : les passagers de Queenstown sont toujours là, mais leur pays vaut `NaN`.
- Conclusion : `left` est plus sûr pour enrichir un tableau, car on ne perd rien et on voit ce qui manque. `inner` est à utiliser en sachant qu'il peut supprimer des lignes.

```python
import pandas as pd
df = pd.read_csv("../exercice-08/data/titanic_propre.csv")
df.shape
```

```python
survie_par_classe_sexe = (df.groupby(["pclass", "sex"])["survived"].mean() * 100).round(1)
survie_par_classe_sexe.unstack()
```

```python
tranches = pd.cut(df["age"], bins=[0, 12, 18, 35, 60, 100],
                  labels=["0-12", "12-18", "18-35", "35-60", "60+"])
survie_par_age = (df.groupby(tranches, observed=True)["survived"].mean() * 100).round(1)
survie_par_age
```

```python
df.groupby(tranches, observed=True)["survived"].count()
```

```python
prix_moyen_par_port = df.groupby("embark_town")["fare"].mean().round(2).sort_values(ascending=False)
prix_moyen_par_port
```

```python
classes_par_port = (pd.crosstab(df["embark_town"], df["pclass"], normalize="index") * 100).round(1)
classes_par_port
```

```python
df.groupby(["pclass", "embark_town"])["fare"].mean().unstack().round(2)
```

```python
survie_seul = (df.groupby("alone")["survived"].mean() * 100).round(1)
survie_seul
```

```python
(df.groupby(["sex", "alone"])["survived"].mean() * 100).round(1)
```

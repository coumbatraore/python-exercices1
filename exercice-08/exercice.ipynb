```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
df = titanic.copy()
```

```python
mediane = df.groupby(["pclass", "sex"])["age"].transform("median")
df["age"] = df["age"].fillna(mediane)
```

```python
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
```

```python
titanic.duplicated().sum()
```

```python
df = df.drop(columns=["deck", "alive", "class", "embarked"])
```

```python
df["survived"] = df["survived"].astype(bool)
df["pclass"] = df["pclass"].astype("category")
df["sex"] = df["sex"].astype("category")
df["embark_town"] = df["embark_town"].astype("category")
```

```python
print(df.shape)
print(df.isna().sum().sum())
df.dtypes
```

```python
q1 = df["fare"].quantile(0.25)
q3 = df["fare"].quantile(0.75)
borne = q3 + 1.5 * (q3 - q1)
hors_bornes = df[df["fare"] > borne]
len(hors_bornes)
```

```python
df.to_csv("data/titanic_propre.csv", index=False)
```
